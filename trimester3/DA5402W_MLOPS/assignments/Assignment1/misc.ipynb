```python
from kafka import KafkaConsumer
consumer = KafkaConsumer(
    'sensor_DA25M622',
    bootstrap_servers=['localhost:9092'],
    auto_offset_reset='earliest',
    enable_auto_commit=True      
)

# simple way - but it's infinite loop (it will forever wait for new messages)
# for msg in consumer:
#     print (msg)

consumed_messages = 0
timeout_ms = 1000   # 1 second
while True:
    msg_pack = consumer.poll(timeout_ms=timeout_ms)   # returns a dict: {TopicPartition: [ConsumerRecord, ...]}
    if not msg_pack:
        print("No more messages found. Exiting loop.")
        break
        
    for tp, messages in msg_pack.items():
        for msg in messages:
            print(msg)
            consumed_messages += 1

#consumer.close()
```
