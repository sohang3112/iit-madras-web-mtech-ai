
# 20. Mini Project

## Task

Design an Airflow pipeline that:

1. Generates synthetic customer data
2. Stores data in CSV
3. Reads CSV
4. Converts data to JSON
5. Produces a summary report

Draw the DAG structure and explain each task.


```python
import faker as faker
import pandas as pd

fake = faker.Faker()
customer_records = []

for _ in range(1000):
    customer_records.append({
        "customer_id": fake.uuid4(),
        "name": fake.name(),
        "age": fake.random_int(18,70),
        "email": fake.email(),
        "state": fake.state()
    })

customer_df = pd.DataFrame(customer_records)

customer_df.to_csv(
    "customer_data.csv",
    index=False
)
```
