```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df1 = pd.read_csv('synthetic_data.csv')
df2 = pd.read_csv('tips.csv')
df2
```

### MCQs and Numerical


##### 1

Create a bar plot using seaborn.barplot that shows the average (mean) value_2 for each category. Based on the plot, what is the approximate average value_2 for category ’A’?


```python
cat_avg_val2s = df1.groupby('category')['value_2'].mean()
sns.barplot(cat_avg_val2s)
plt.title('Average value_2 for each category')
plt.show()
print(cat_avg_val2s['A'])
```

#### 2

Create a scatter plot using seaborn.scatterplot with value_1 on the x-axis and value_2 on the y-axis. What kind of relationship do value_1 and value_2 appear to have?

My answer: (based on below plot) Strong Negative Correlation


```python
sns.scatterplot(df1, x='value_1', y='value_2')
```

#### 3

Create a box plot using seaborn.boxplot showing value_1 for each category. One category shows an outlier beyond the whisker. Which category is it?

My Answer: C (since category C has outlier shown as hollow circle far from top).

NOTE: box plot outliers are values lying > 1.5x IQR (Inter-Quartile Range) away from 1st or 3rd quartile.


```python
sns.boxplot(df1, x='category', y='value_1')
```

#### 4

Create a histogram of value_2 with bins [20, 30, 40, 50, 60]. How many points fall between 20 and 30?

My Answer: 3 (from below histogram).

NOTE: while plotting historgram with bins, we omitted 0-20, so 4 / 20 total df length is missing.


```python
sns.histplot(df1, x='value_2', bins=[20,30,40,50,60])
```

#### 5

Create a count plot using seaborn.countplot for the day column. Which day has the highest count?

My Answer: (based on count plot) Sat


```python
sns.countplot(df2, x='day')
```

#### 6

Create a box plot for the total_bill column. What is the approximate median total_bill (nearest dollar)?


```python
sns.boxplot(df2, x='total_bill')
plt.show()
print(df2['total_bill'].median())
```

#### 7

Create a bar plot showing the average tip for each sex. Which group has a higher average tip?

My Answer (from bar plot): Male


```python
avg_sex_tip = df2.groupby('sex')['tip'].mean()
sns.barplot(avg_sex_tip)
```

#### 8

Create a scatter plot of total_bill vs tip, with hue=’time’. Which group (Lunch or Dinner) has the point with both highest total_bill and tip?

My Answer (from scatter plot): Dinner


```python
sns.scatterplot(df2, x='total_bill', y='tip', hue='time')
```
