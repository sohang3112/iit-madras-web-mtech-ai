### Slide 3: Lambda Function


```python
# Named function
def square(x):
    return x**2

square(5)
```

```python
# Equivalent lambda
square_lambda = lambda x: x**2
a = square_lambda(5)
a

# Note square_lambda is just a variable to which lambda is assigned but there is no function named as square_lambda here
```

```python
# Lambda can be used directly without being assigned to a variable
y=10
(lambda x,y: x**2+y**2)(5,10)
```

```python
# Filtering a list based on a condition within a lambda function
nums = [3, 5, 8, 10]
list(filter(lambda x: x % 2 == 0, nums))
```

```python
# Map each element in list to its square using lambda function within a map function
list(map(lambda x: x**2, nums))
```

```python
# To check if a number is even or odd
check_even_odd = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check_even_odd(4))
print(check_even_odd(7))
```

```python
# Looping over a list and apply square function over all numbers
nums = [-3, -1, 0, 2, 5, -7]
Eo = [(lambda x: "Even" if x%2==0 else "Odd")(x) for x in nums]
print(Eo)
```

### Slide 6 : Selection using loc


```python
df = pd.DataFrame({
    "Name": ["Asha", "Bala", "Chetan", "Divya"],
    "Marks": [85, 90, 78, 92],
    "City": ["Chennai", "Pune", "Delhi", "Vizag"]
}, index=["s1", "s2", "s3", "s4"])
df
```

```python
# Access multiple rows & columns
df.loc[["s1","s2"], ["Name","Marks"]]
```

```python
# Conditional selection
df.loc[(df["Marks"] > 80) & (df["City"]=="Chennai"), ["Name", "City"]]
```

### Slide 14 : Updating values in DataFrame


```python
df = pd.DataFrame({
    "Name": ["Asha", "Bala", "Chetan", "Divya", "Jay"],
    "Marks": [85, 40, 78, 92, 53],
    "City": ["Chennai", "Pune", "Delhi", "Vizag", "Mumbai"]
})
```

```python
# Set a single cell (label-based)
df.loc[1, "Marks"] = 95 # 1 here refers to index and not position though both are same
df
```

```python
# NEW TO ME: Set a single cell (label-based)
df.at[1, "Marks"] = 90 # Faster than .loc for scalar assignment
df
```

```python
# Set via position
df.iloc[0, 1] = 88
df
```

```python
# NEW TO ME: Set via position
df.iat[0, 1] = 90 # Faster than .iloc for scalar assignment
df
```

```python
# Set multiple cells via a condition and broadcasting
df.loc[df["City"] == "Delhi", "Marks"] = 80
df
```

```python
# Set entire column via broadcasting
df["Passed"] = (df["Marks"] >= 80)
df
```

```python
# Setting conditionally via a lambda function: Increase marks by 5 for those below 60
# Vectorised and Broadcasted and inplace modification
df.loc[lambda x: x["Marks"] < 60, "Marks"] += 5
df
```

```python
# NEW TO ME: Conditional setting via .where()
# .where() keeps values where a condition is True, replaces values where the condition is False and returns a new object
df1 = df.copy()
df1['City'].where(df1['Marks'] == 90, inplace=True) # NaN is the default replacement
df1
```

```python
# Value substitution using replace
# Can also be used with regex
df = pd.DataFrame({
    "Name": ["Asha", "Bala", "Chetan", "Divya", "Jay"],
    "Marks": [85, 40, 78, 92, 53],
    "City": ["Chn", "Pune", "Delhi", "Hyd", "Mumbai"]
})
df['City'] = df['City'].replace({'': 'Hyderabad', 'Chn': 'Chennai'})
df
```

### Slide 17: _Grouping_ Operations


```python
df = pd.DataFrame({
    "Department": ["Sales", "Sales", "HR", "HR", "IT", "IT"],
    "Employee": ["Asha", "Bala", "Chetan", "Divya", "Esha", "Farhan"],
    "Salary": [50000, 55000, 40000, 42000, 60000, 62000],
    "Experience": [2, 4, 3, 5, 6, 8]
})
print(df)
grouped = df.groupby("Department")
grouped  # GroupBy object and not a dataframe
```

```python
grouped.groups
```

```python
# NEW TO ME
grouped.get_group("HR")
```

### Slide 17: Aggregations after Grouping


```python
# Simple aggregation resulting in creation of a new series object
df1 = df.groupby("Department")["Salary"].mean()
df1
```

```python
# Multiple aggregations
df.groupby("Department")["Salary"].agg(["mean", "max", "min"])
```

```python
# Different functions for different columns
df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Experience": "median"
})
```

```python
# Group by and Aggregate by more than one column
df.groupby(["Department", "Experience"])["Salary"].mean()
```

```python
# Custom function based aggregation
def range(x):
    return x.max() - x.min()

df.groupby("Department")["Salary"].agg(["mean", range])
```

### Slide 18: Transform and Filter after Grouping


```python
df = pd.DataFrame({
    "Department": ["Sales", "Sales", "HR", "HR", "IT", "IT"],
    "Employee": ["Asha", "Bala", "Chetan", "Divya", "Esha", "Farhan"],
    "Salary": [50, 55, 40, 42, 60, 62]
})

# NEW TO ME
df["DeptAvgSalary"] = df.groupby("Department")["Salary"].transform("mean") # Multiple transformations are not allowed at once in recent versions of pandas
df
```

```python
# Keep only departments where average salary > 50
df_filtered = df.groupby("Department").filter(lambda x: x["Salary"].mean() > 50)
df_filtered
```

```python
# Filter by presence of a condition
# Keeps any department that has at least one employee earning > 60
df.groupby("Department").filter(lambda g: (g["Salary"] > 60).any())
```

```python
# Filter by all condition
# Keep only departments where all salaries are above 45
df.groupby("Department").filter(lambda g: (g["Salary"] > 52).all())
```

### Slide 20 : Categorical Values in Pandas


```python
df = pd.DataFrame({
    "Region": ["East", "West", "North", "East", "West"]
})
print(df["Region"].dtype)
df["Region"] = df["Region"].astype("category") # Setting the data type of the column to "category"
print(df["Region"].dtype)
```

```python
df["Region"].cat.categories  # Returns the list of categories used in a column
```

```python
# Rename categories
df["Region"] = df["Region"].cat.rename_categories({"East": "E", "West": "W", "North": "N"})

# Reorder (for ordered comparisons)
df["Region"] = df["Region"].cat.reorder_categories(["N", "E", "W"], ordered=True)

# Add Category
df["Region"] = df["Region"].cat.add_categories(["S"])
```

```python
df["Region"].cat.categories
```
