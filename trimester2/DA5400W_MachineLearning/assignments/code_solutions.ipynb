```python
# Q10 > 1.2 
from plotly import graph_objects as go
import numpy as np
x = np.linspace(-50, 50, num=200)
y = np.linspace(-50, 50, num=200)
z = x + y
#z = z[x**2 + y**2 - 1 == 0]
fig = go.Figure(data=[go.Surface(x=x, y=y, z=z)])
fig.show()
```

```python
# Q10 > 3
from plotly import graph_objects as go
import numpy as np
x = np.linspace(-50, 50, num=200)
y = np.linspace(-50, 50, num=200)
z = (x - 1)**2 + y**2
#z = z[x - y**2 <= 0]
fig = go.Figure(data=[go.Surface(x=x, y=y, z=z)])
fig.show()
```
