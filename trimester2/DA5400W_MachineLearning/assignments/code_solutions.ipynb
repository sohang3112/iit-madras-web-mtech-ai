```python
import numpy as np
roots = np.roots([-1, 4, 1, -10, 4, -8])
roots
```

```python
f = lambda x: -x**5 + 4*x**4 + x**3 - 10*x**2 + 4*x - 8
#fapprox = lambda x: (x-2)*(x-3.5)*(x+1.7)*(x-0.1-0.8j)*(x-0.1+0.8j)
fapprox = lambda x: np.prod(x-roots)
from pprint import pprint
pprint({x: f(x) for x in roots})
for x in range(0, 21):
    print(x, fapprox(x) - f(x))
```
