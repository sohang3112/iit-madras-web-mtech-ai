Generate 100 training samples: $y = x + 0.3 x^2 + 0.7 x^3 + w$ :

- Uniformly distributed: $x \in [-3, 3]$
- $w$ is Gaussian noise with mean 0, standard deviation 0.09

Perform linear regression for model sizes M = 1 to 120 and plot:

- Bias (average of $y - f(x)$) vs $M$
- Variance of ~~y~~ $f(x)$ vs $M$
- Training error vs $M$
- Validation error vs $M$


```python
import numpy as np
```

```python
def data(size: int):    
    x = np.random.uniform(-3, 3, size)
    w = np.random.normal(0, 3, size)  # noise
    y = x + 0.3 * x**2 + 0.7 * x**3 + w
    return x, y 

xtrain, ytrain = data(size=100)
xval, yval = data(size=25)

biases, variances, train_errors, val_errors = [], [], [], []
for M in range(1, 121):
    coeffs = np.polyfit(xtrain, ytrain, deg=M)
    model = np.poly1d(coeffs)
    ytrain_pred = model(xtrain)
    yval_pred = model(xval)
    biases.append(np.mean(yval - yval_pred))
    variances.append(np.var(yval_pred))
    train_errors.append(np.mean((ytrain - ytrain_pred) ** 2))
    val_errors.append(np.mean((yval - yval_pred) ** 2))
    
```
